```sql
-- ============================================================
-- KRAF | FASE 2 | CONSULTA OLAP 04
-- Contexto económico
-- Responsable: Fernando Boveda Pleitez
--
-- Pregunta de negocio:
-- ¿Cómo se comporta el contexto económico disponible para
-- El Salvador y qué información externa puede complementar
-- el análisis de KRAF?
--
-- Objetivo:
-- Consultar inflación y calificación soberana disponibles
-- para El Salvador.
--
-- Nota:
-- La fuente económica disponible para El Salvador llega
-- hasta 2023.
-- ============================================================

SELECT
    country_code,
    country_name,
    year,
    ROUND(inflation_rate, 2) AS inflation_rate,
    rating_foreign,
    rating_foreign_outlook,
    rating_local,
    rating_local_outlook
FROM workspace.kraf_gold.contexto_economico_pais
WHERE country_code = 'SLV'
ORDER BY year DESC;
```
