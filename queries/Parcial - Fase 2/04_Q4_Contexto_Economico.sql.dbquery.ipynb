```sql
-- ============================================================
-- KRAF | FASE 2 | CONSULTA OLAP 04
-- Contexto económico
-- Responsable: Fernando Boveda Pleitez
--
-- Pregunta de negocio:
-- ¿Cómo se comporta la inflación y la calificación crediticia
-- de los países disponibles?
--
-- Objetivo:
-- Analizar el contexto económico disponible para complementar el análisis de KRAF.
-- ============================================================

SELECT
    country_name,
    year,
    ROUND(inflation_rate, 2) AS inflation_rate,
    rating_foreign,
    rating_foreign_outlook
FROM workspace.kraf_gold.contexto_economico_pais
WHERE inflation_rate IS NOT NULL
ORDER BY year DESC, inflation_rate DESC;
```
