```sql
-- ============================================================
-- KRAF | FASE 2 | CONSULTA OLAP 06
-- Variación del Tipo de Cambio
-- Responsable: Karen María Mina Martínez
--
-- Pregunta de negocio:
-- ¿Qué monedas latinoamericanas presentan mayores variaciones
-- en el periodo analizado?
--
-- Objetivo:
-- Comparar la variación entre los valores iniciales y finales de cada
-- moneda disponible en el contexto cambiario.
-- ============================================================

WITH valores AS (

    SELECT
        MIN(fecha) AS fecha_inicial,
        MAX(fecha) AS fecha_final
    FROM workspace.kraf_gold.contexto_tipo_cambio
),

tipo_cambio AS (

    SELECT
        'Real brasileño' AS moneda,
        MIN(brazilian_real) AS valor_inicial,
        MAX(brazilian_real) AS valor_final
    FROM workspace.kraf_gold.contexto_tipo_cambio

    UNION ALL

    SELECT
        'Peso chileno',
        MIN(chilean_peso),
        MAX(chilean_peso)
    FROM workspace.kraf_gold.contexto_tipo_cambio

    UNION ALL

    SELECT
        'Peso colombiano',
        MIN(colombian_peso),
        MAX(colombian_peso)
    FROM workspace.kraf_gold.contexto_tipo_cambio

    UNION ALL

    SELECT
        'Peso mexicano',
        MIN(mexican_peso),
        MAX(mexican_peso)
    FROM workspace.kraf_gold.contexto_tipo_cambio
)

SELECT
    moneda,
    ROUND(valor_inicial, 4) AS valor_inicial,
    ROUND(valor_final, 4) AS valor_final,
    ROUND(valor_final - valor_inicial, 4) AS variacion_absoluta,
    ROUND(
        ((valor_final - valor_inicial) / NULLIF(valor_inicial, 0)) * 100,
        2
    ) AS variacion_porcentual

FROM tipo_cambio

ORDER BY ABS(valor_final - valor_inicial) DESC;
```
