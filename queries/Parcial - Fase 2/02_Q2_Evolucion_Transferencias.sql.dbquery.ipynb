```sql
-- ============================================================
-- KRAF | FASE 2 | CONSULTA OLAP 02
-- Evolución temporal de transferencias
-- Responsable: Rodrigo Javier Villalobos Hernández
--
-- Pregunta de negocio:
-- ¿Cómo cambia la actividad de transferencias de KRAF
-- durante el año?
--
-- Objetivo:
-- Comparar la cantidad de transferencias y los montos
-- registrados por mes.
-- ============================================================

SELECT
    YEAR(fecha_hora) AS anio,
    MONTH(fecha_hora) AS mes,
    COUNT(*) AS total_transferencias,
    ROUND(SUM(monto), 2) AS monto_total,
    ROUND(AVG(monto), 2) AS monto_promedio
FROM workspace.kraf_silver.transferencia_silver
GROUP BY YEAR(fecha_hora), MONTH(fecha_hora)
ORDER BY anio, mes;
```
