```sql
-- ============================================================
-- KRAF | FASE 2 | CONSULTA OLAP 05
-- Segmentación sociodemográfica
-- Responsable: Rodrigo Javier Villalobos Hernández
--
-- Pregunta de negocio:
-- ¿Cómo se distribuye la población sintética integrada en KRAF
-- según departamento y área?
--
-- Objetivo:
-- Identificar los segmentos con mayor cantidad de personas y
-- analizar la edad promedio entre áreas y departamentos de El Salvador.
--
-- Nota:
-- Nemotron es un dataset sintético y no representa clientes
-- reales de KRAF.
-- ============================================================

SELECT
    department,
    area,
    total_personas,
    ROUND(edad_promedio, 1) AS edad_promedio,
    edad_minima,
    edad_maxima
FROM workspace.kraf_gold.contexto_sociodemografico_sv
ORDER BY total_personas DESC;
```
