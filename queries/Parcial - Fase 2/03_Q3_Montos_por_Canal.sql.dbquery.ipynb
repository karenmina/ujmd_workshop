```sql
-- ============================================================
-- KRAF | FASE 2 | CONSULTA OLAP 03
-- Comportamiento de montos por canal
-- Responsable: Alejandro Raphael Mejía Guerrero
--
-- Pregunta de negocio:
-- ¿Cómo se distribuyen las transferencias según el canal?
--
-- Objetivo:
-- Comparar el comportamiento de los montos según el canal utilizado.
-- ============================================================

SELECT
    canal,
    COUNT(*) AS total_transferencias,
    ROUND(SUM(monto), 2) AS monto_total,
    ROUND(AVG(monto), 2) AS monto_promedio,
    ROUND(MIN(monto), 2) AS monto_minimo,
    ROUND(MAX(monto), 2) AS monto_maximo
FROM workspace.kraf_silver.transferencia_silver
GROUP BY canal
ORDER BY monto_total DESC;
```
