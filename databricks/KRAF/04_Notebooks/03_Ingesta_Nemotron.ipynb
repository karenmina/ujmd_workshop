```python
from huggingface_hub import hf_hub_download
import pandas as pd

path1 = path

path2 = hf_hub_download(
    repo_id="nvidia/Nemotron-Personas-El-Salvador",
    repo_type="dataset",
    filename="data/train-00001-of-00003.parquet"
)

path3 = hf_hub_download(
    repo_id="nvidia/Nemotron-Personas-El-Salvador",
    repo_type="dataset",
    filename="data/train-00002-of-00003.parquet"
)

df1 = pd.read_parquet(path1)
df2 = pd.read_parquet(path2)
df3 = pd.read_parquet(path3)

df = pd.concat([df1, df2, df3], ignore_index=True)

print("Registros totales:", len(df))
print("Columnas totales:", len(df.columns))
```
